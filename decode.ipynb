```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month_name()
df['Day'] = df['Date'].dt.day_name()
```

```python
df.head()
```

```python
df['Product'] = df['Product'].str.strip()

df['PaymentMethod'] = df['PaymentMethod'].str.strip()
```

```python
df['OrderStatus'] = df['OrderStatus'].str.strip()

df['ReferralSource'] = df['ReferralSource'].str.strip()
```

```python
#project-2
```

```python
total_orders = df['OrderID'].count()
print("Total Orders:", total_orders)
```

```python
print("Average Quantity:", round(df['Quantity'].mean()))
print("Average Unit Price:", round(df['UnitPrice'].mean(),2))
print("Average Total Price:", round(df['TotalPrice'].mean(),2))
```

```python
print("Median Quantity:", round(df['Quantity'].median()))
print("Median Unit Price:", df['UnitPrice'].median())
print("Median Total Price:", df['TotalPrice'].median())
```

```python
sns.histplot(df['Quantity'], bins=10)
plt.title("Distribution of Quantity")
plt.show()
```

```python
sns.histplot(df['TotalPrice'], bins=20)
plt.title("Distribution of Total Price")
plt.show()
```

```python
sns.countplot(data=df, x='PaymentMethod')
plt.show()
```

```python
monthly_sales = df.groupby('Month')['TotalPrice'].sum()

monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Revenue Trend")
plt.ylabel("Revenue")
plt.show()
```

```python
product_sales = df.groupby('Product')['Quantity'].sum()

product_sales.sort_values(ascending=False).plot(kind='bar')
plt.title("Top Selling Products")
plt.show()
```

```python
sns.boxplot(x=df['TotalPrice'])
plt.title("Outlier Detection - Total Price")
plt.show()
```

```python
sns.boxplot(x=df['Quantity'])
plt.title("Outlier Detection - Quantity")
plt.show()
```
